###For Exercise 1-3 below,  find a hyperplane in $\mathbb{R}^n$ containing the vector $\boldsymbol x_0$, listed first, and whose normal vector is $\boldsymbol n$, listed second.


---


*An example is provided in Exercise 0. To execute your code, hit the play/execute/run button or Ctrl+Shift.*

0. $\begin{bmatrix} 1\\2\\3\end{bmatrix},\begin{bmatrix} 3\\-2\\0\end{bmatrix}$

```python
import numpy as np

x = np.array([1,2,3])
n = np.array([3,-2,0])

#3(x_0-1) - 2(x_1-2) + 0(x_2-3) = 0
#3x_0 - 2x_1 = -1
print(' + '.join([f'{n[i]}x_{i}' for i in range(len(n))]) + f' = {x@n}')
```

1. $\begin{bmatrix}0\\0\\1\\2\end{bmatrix},\begin{bmatrix}1\\-2\\2\\-1\end{bmatrix}$


```python
### EXERCISE 1#######################################
x = np.array([0,0,1,2])
n = np.array([1,-2,2,-1])

print('solution') #compute the hyperplane here
```

<details><summary>Click here for the solution</summary>

$x_0-2x_1+2x_2-x_3=0$

</details>

```python
### EXERCISE 2#######################################
x = np.array([1,2,3,4,5])
n = np.array([-3,-6,3,2,-5])

print('solution') #compute the hyperplane here
```

3. $\begin{bmatrix}10\\-8\\6\\-4\\2\end{bmatrix},\begin{bmatrix}1\\3\\5\\7\\9\end{bmatrix}$

###For Exercise 4-7 below, use inner products to express the vector $\boldsymbol y$, listed first, as a linear combination of the orthogonal basis $\mathcal B$, listed second.

*An example is provided in Exercise 0. To execute your code, hit the play/execute/run button or Ctrl+Shift.*

0. $\begin{bmatrix}-2\\-6\end{bmatrix}$, $\left\{\begin{bmatrix}1\\3\end{bmatrix}\right\}$

```python
### EXERCISE 0#######################################
y = np.array([-2,-6])
B = np.array([[1,3]])

#fourier = (B[0]@y)/(B[0]@B[0])
#y = (fourier)*B[0]
fourier = np.array([(b@y)/(b@b) for b in B])
print(f'y = ' + ' + '.join([f'{fourier[i]}v_{i}' for i in range(len(B))]))
```

4. $\begin{bmatrix}-1\\10\\11\end{bmatrix}$, $\left\{\begin{bmatrix}1\\2\\1\end{bmatrix},\begin{bmatrix}1\\0\\-1\end{bmatrix}\right\}$


```python
### EXERCISE 4#######################################
y = np.array([-1,10,11])
B = np.array([[1,2,1],
              [1,0,-1]])

print('solution') #write combination here
```
