###For Exercises 3-6 below, determine if $\boldsymbol b$ is a linear combination of the other vectors $\{\boldsymbol a_1, \boldsymbol a_2, \ldots\}$. If so, write $\boldsymbol b$ as a linear combination. The set of vectors $\{\boldsymbol a_1, \boldsymbol a_2,\ldots\}$ will be listed first, followed by the vector $\boldsymbol b$. Answers may vary.
---


*An example is provided in Exercise 0.

0.  $\left\{ \begin{bmatrix}1\\2\end{bmatrix},\begin{bmatrix}3\\-1\end{bmatrix}\right\},\ \begin{bmatrix}4\\1\end{bmatrix}$


````python
import numpy as np
### EXERCISE 0#######################################
A = np.array([[1,3],[2,-1]])
b = np.array([4,1])

x = np.linalg.lstsq(A,b)[0].round() #the solutions, if any, will be integers, so the rounding helps clean it up

print(f'{np.all(A@x==b)}, b = {x[0]}*a_1 + {x[1]}*a_2')
````

3.  $\left\{ \begin{bmatrix}-1\\3\end{bmatrix},\begin{bmatrix}2\\-2\end{bmatrix}\right\},\ \begin{bmatrix}1\\5\end{bmatrix}$

````python
### EXERCISE 3#######################################
A = np.array([[-1,2],[3,-2]])
b = np.array([1,5])

print('Is b a linear combination of A?')
````

<details><summary>Click here for the solution</summary>

```Python
True
```
$\boldsymbol b = 3\boldsymbol a_1 + 2\boldsymbol a_2$

</details>

4.  $\left\{ \begin{bmatrix}-3\\-1\\1\end{bmatrix},\begin{bmatrix}2\\2\\3\end{bmatrix}\right\},\ \begin{bmatrix}5\\4\\3\end{bmatrix}$

````python
### EXERCISE 4#######################################
A = np.array([[-3,2],[-1,2],[1,3]])
b = np.array([5,4,3])

print('Is b a linear combination of A?')
````

5.  $\left\{ \begin{bmatrix}3\\-2\\1\end{bmatrix},\begin{bmatrix}-2\\3\\-3\end{bmatrix}\right\},\ \begin{bmatrix}1\\6\\-9\end{bmatrix}$

6.  $\left\{ \begin{bmatrix}1\\1\\2\end{bmatrix},\begin{bmatrix}-2\\-1\\-2\end{bmatrix},\begin{bmatrix}-1\\1\\2\end{bmatrix}\right\},\ \begin{bmatrix}-4\\-1\\-2\end{bmatrix}$

````python
### EXERCISE 6#######################################
A = np.array([[1,-2,-1],[1,-1,1],[2,-2,2]])
b = np.array([-4,-1,-2])

print('Is b a linear combination of A?')
````

<details><summary>Click here for the solution</summary>

```Python

```
