0.   $\left[\begin{array}{rr|r}3&0&4\\4&1&-1\\8&3&-2\end{array}\right]$


```python
import numpy as np

A = np.array([[3,0],
              [4,1],
              [8,3]])
b = np.array([4,-1,-2])

x = np.linalg.lstsq(A,b)[0]
print(x)
```
