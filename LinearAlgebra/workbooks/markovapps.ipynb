1.   In the decade 2010-2020, the US gained on average 1,063,000 immigrants per year. Assuming the distribution of ages for immigrants entering the US was the same as those residents already living in the US, estimate the 2020 US population using the same data from above (Example 3.8.2 in the textbook). Estimate the US 2020 population using a linear dynamical system which incorporates this immigration. How many 20 year-olds would the US have in 2020?


````python
### EXERCISE 1#######################################
year=2020
x = pop

plt.plot(x)
for i in range(year-2010):
    x = A @ x        # The dynamic equation is the only line to be modified for immigration
plt.plot(x, "--")
plt.xlabel('Age')
plt.ylabel('Population (millions)')
plt.legend(['2010', str(year)])
plt.show()

print("2010 Population:", round(sum(pop),1), "million")
print("Predicted", year, "Population:", round(sum(x),1), "million")
print("Actual 2020 Population: 329.5 million")
print(x[20])
````

2.   Imagine in 2010 that zombie infestation in the USA begins to spread. Given the same birth and death rates as before, assume everyone that dies, turns into a zombie. Assume all humans in the USA, that do not die from other means, are additionally at a 10% risk of becoming a zombie each year by infection. Zombies do not ever become human again and humans manage to kill about 5% of the zombies each year. Ignore immigration (because who wants to move next to a zombie). The initial population of zombies in 2010 was only 10 zombies.


````python
### EXERCISE 2#######################################
year=2020
x = np.concat((pop,[1e-5]))  #initial population
inf = 0.9  #infection rate

# the transition matrix needs a new row/column to accomodate for zombies
# it will be the only section of this code that needs to be modified
# as all other adjustments have been made for you
A = np.vstack((b,
              np.block([np.diag(1-d[:99]), np.zeros((99,2))]),
              (1-d[99])*np.eye(1,101,99) + (1-d[100])*np.eye(1,101,100)))


plt.plot(x[:-1])
for i in range(year-2010):
    x = A @ x
plt.plot(x[:-1], "--")
plt.xlabel('Age')
plt.ylabel('Population (millions)')
plt.legend(['2010', str(year)])
plt.show()

print("2010 Human Population:", round(sum(pop),1), "million")
print("2010 Zombie Population: 10")
print("Predicted", year, "Humman Population:", round(sum(x[:-1]),1), "million")
print("Predicted", year, "Zombie Population:", round(x[-1],1), "million")
````
