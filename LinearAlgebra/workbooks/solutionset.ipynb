###For Exercises 1-2 below, let $A$ be the provided matrix. Find a basis for $\text{col}(A)$ consisting of column vectors of $A$ and a basis for $\text{nul}(A)$. Find the rank and nullity of $A$. Answers may vary.

0.   $\begin{bmatrix}1&1&1&0\\2&3&4&2\\3&1&3&1\end{bmatrix}$


```python
import numpy as np

def row_scaling(coefficient_matrix, scalar, i): #perform the elementary row operation of scaling
    A = coefficient_matrix.astype(float)
    A[i] = scalar*coefficient_matrix[i]

    return A

def row_replacement(coefficient_matrix, scalar, i, j): #perform the elementary row operation of replacement
    A = coefficient_matrix.astype(float)
    A[i] = coefficient_matrix[i] + scalar*coefficient_matrix[j]

    return A

def row_interchange(coefficient_matrix, i, j): #perform the elementary row operation of interchange
    idx = np.arange(len(coefficient_matrix))
    idx[i],idx[j]  = j,i

    return coefficient_matrix[idx]

A = np.array([[1,1,1,0],
              [2,3,4,2],
              [3,1,3,1]])
#Using the Row Operations above, we compute the RREF
RREF = row_replacement(A,-2,1,0)
RREF = row_replacement(RREF,-3,2,0)
RREF = row_replacement(RREF,2,2,1)
RREF = row_scaling(RREF,0.25,2)
RREF = row_replacement(RREF,-1,0,2)
RREF = row_replacement(RREF,-2,1,2)
RREF = row_replacement(RREF,-1,0,1)
print(RREF,'\n')

#Using the RREF, we recognize the pivots are columns 0,1,2
print('rank = 3,','col(A) = Span(', A[:,0],A[:,1],A[:,2],')\n')
print('nullity = 1,', 'nul(A) = Span(', np.array([3,2,-5,4]),')')
```
