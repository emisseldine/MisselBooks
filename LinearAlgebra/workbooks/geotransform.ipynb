0.   Reflect across the $y$-axis, stretch by a factor of 2 in the $x$-direction, then expand by a factor of $3$ in the $y$-direction.


```python
import numpy as np

dilate = lambda a, b: np.array([[a, 0], [0, b]]) #stretch horizontally by a and vertically by b
rotate = lambda theta: np.array([[np.cos(theta), -np.sin(theta)],[np.sin(theta), np.cos(theta)]]) #rotate about the origin by theta
reflect = lambda theta: np.array([[np.cos(2*theta), np.sin(2*theta)],[np.sin(2*theta), -np.cos(2*theta)]]) #reflect across the line y = tan(theta)*x
shear = lambda theta, m: 1/2*np.array([[2-m*np.sin(2*theta), m+m*np.cos(2*theta)],[-m+m*np.cos(2*theta), 2+m*np.sin(2*theta)]])
project = lambda theta: 1/2*np.array([[1+np.cos(2*theta), np.sin(2*theta)],[np.sin(2*theta), 1-np.cos(2*theta)]])

A = dilate(2,3) @ reflect(np.pi/2)
print(A)

import matplotlib.pyplot as plt
plt.figure(figsize=(8, 8), dpi=80)
plt.axis([-2,1,0,3])

#create a scatter plot for the "Unit Square"
points = np.concatenate([[[0.01*i,0.01*j] for i in range(101)] for j in range(101)])

#Transform the set of points
transformed = np.array([A @ x for x in points])

#Show the sets of points
plt.scatter([x[0] for x in points], [x[1] for x in points])
plt.scatter([x[0] for x in transformed], [x[1] for x in transformed])

plt.show()
```

7. Find a vector $\boldsymbol x\in \mathbb R^3$ such that $T(\boldsymbol x) = (-6,8,12)$.

```python
### EXERCISE 7#######################################
def row_scaling(coefficient_matrix, scalar, i):  #Solve the linear system Ax + b = T(x) using row operations
    A = coefficient_matrix.astype(float)
    A[i] = scalar*coefficient_matrix[i]
    return A
def row_replacement(coefficient_matrix, scalar, i, j):
    A = coefficient_matrix.astype(float)
    A[i] = coefficient_matrix[i] + scalar*coefficient_matrix[j]
    return A
def row_interchange(coefficient_matrix, i, j):
    idx = np.arange(len(coefficient_matrix))
    idx[i] = j
    idx[j] = i
    return coefficient_matrix[idx]
def zero_below(coefficient_matrix, pivot):
    A = coefficient_matrix.astype(float)
    if abs(A[pivot]) < 1e-6:
        for k in range(pivot[0]+1,len(A)):
            if abs(A[k,pivot[1]]) > 1e-6:
                A = row_interchange(A,k,pivot[0])
                break
    for k in range(pivot[0]+1,len(A)):
        A = row_replacement(A,-A[k,pivot[1]]/A[pivot],k,pivot[0])
    return A
def zero_above(coefficient_matrix, pivot):
    A = coefficient_matrix.astype(float)
    A = row_scaling(A,1/A[pivot],pivot[0])
    for k in range(pivot[0]):
        A = row_replacement(A,-A[k,pivot[1]],k,pivot[0])
    return A

```
