0.   $\left\{\begin{bmatrix}1\\-1\\3\end{bmatrix},\begin{bmatrix}2\\-2\\6\end{bmatrix},\begin{bmatrix}-2\\3\\-2\end{bmatrix}\right\}$

```python
import numpy as np

A = np.column_stack([[1,-1,3],
              [2,-2,6],
              [-2,3,-2]])
if len(A)==np.linalg.matrix_rank(A): #does number of pivots equal columns?
  print('linearly independent')
else:
  print('linearly dependent')

Asub = A[:,[0,2]]  # Is v[1] a linear combination of v[0] and v[2]?
x = np.linalg.lstsq(Asub,A[:,1])[0] #solve the linear system
print(Asub@x)                       #notice this product is 1*v[1]
print(np.round(x,0),'\n')           #2*v[0] + 0*v[2] = 1*v[1]
print('2*v[0]-v[1]+0*v[2]=np.zeros(3)')
```
