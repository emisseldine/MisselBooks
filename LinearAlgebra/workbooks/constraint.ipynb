###For Exercises 1-4 below, solve the constrained least squares problem $A\boldsymbol x = \boldsymbol b$ subject to $C\boldsymbol x=\boldsymbol d$.


---


*An example is provided in Exercise 0. To execute your code, hit the play/execute/run button or Shift+Enter.*


0.   $A = \begin{bmatrix}1&1\\0&2\end{bmatrix}$, $\boldsymbol b = \begin{bmatrix}4\\3\end{bmatrix}$, $C = \begin{bmatrix}1&-1\end{bmatrix}$, $\boldsymbol d = \begin{bmatrix}-1\end{bmatrix}$





````python
import numpy as np

def lstsq_constraint(A, b, C, d):
    ATA = A.T @ A
    CT = (C.T).reshape((len(ATA), -1))
    n = len(CT[0])
    KKT = np.block([[2*ATA, CT], [CT.T, np.zeros((n, n))]])
    KKTb = np.hstack((2*A.T @ b, d))
    return np.linalg.lstsq(KKT, KKTb, rcond=None)[0][:-n]

A = np.array([[1,1],[0,2]])
b = np.array([4,3])
C = np.array([1,-1])
d = -1

print(lstsq_constraint(A,b,C,d))
````
