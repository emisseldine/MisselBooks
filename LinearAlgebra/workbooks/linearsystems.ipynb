0.   Determine if the given points are solutions to the linear system
$\begin{cases}
x~~+y &=\quad\ 7\\
3x+3y-10z &=-79\\
x~\ +5y+~~~~z &=~~~33.
\end{cases}$

i\. $(4,3,10)$\
ii\. $(0,0,0)$\
iii\. $(3,4,10)$\
iv\. $(-10,3,-38)$





````python
import numpy as np

IsLinearSolution = lambda A,b,x : np.array_equal(A@x,b) #Checks if A@x equals b

A = np.array([[1,1,0],[3,3,-10],[1,5,1]])
b = np.array([7,-79,33])

for x in [[4,3,10],[0,0,0],[3,4,10],[-10,3,-38]]:
  print(x,IsLinearSolution(A,b,x))
````
