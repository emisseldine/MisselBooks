###For Exercises 1-2 below, for the given objectives $J_i$, solve the multi-objective least squares problem $$J = \lambda_1J_1 + \lambda_2J_2 + \ldots + \lambda_kJ_k,$$  where $J_i = \Vert A_i \boldsymbol x - \boldsymbol b_i\Vert^2$.


---

0.   $A_1=\begin{bmatrix}2&3\\4&9\end{bmatrix}$, $\boldsymbol b_1 = \begin{bmatrix}5\\11\end{bmatrix}$, $\lambda_1=1$;

 $A_2=\begin{bmatrix}1&13\end{bmatrix}$, $\boldsymbol b_2 = \begin{bmatrix}\dfrac{19}{3}\end{bmatrix}$, $\lambda_2=\dfrac{1}{4}$


````python
import numpy as np

def multi_lstsq(As, bs, lambdas):
    k = len(lambdas)
    A = np.concatenate([np.sqrt(lambdas[i])*np.array(As[i]) for i in range(k)])
    b = np.concatenate([np.sqrt(lambdas[i])*np.array(bs[i]) for i in range(k)])
    return np.linalg.lstsq(A,b, rcond=None)[0]

# For each of the following, solve the multi-objective least squares problem
# J = lambda_1J_1 + lambda_2J_2 + ... + lambda_kJ_k
# where J_i = ||A_i @ x - b_i||**2

# J_1
A1 = np.array([[2,3],
              [4,9]])
b1 = np.array([5,11])
lambda1 = 1

# J_2
A2 = np.array([[1,13]])
b2 = np.array([19/3])
lambda2 = 1/4

As = [A1,A2]
bs = [b1,b2]
lambdas = [lambda1,lambda2]
x = multi_lstsq(As,bs,lambdas)
print(x)
````
