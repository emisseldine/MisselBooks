0.   $\begin{bmatrix}
1&2\\
-4&5
\end{bmatrix}$





````python
import numpy as np

A = np.array([[1,2],[-4,5]])
print(A,'\n')

# replace Row 2 with Row 2 + 4 Row 1
A[1] = A[1] + 4*A[0]
print(A, '\n replace Row 2 with Row 2 + 4 Row 1\n echelon form\n')

# scale Row 2 by 1/13
A[1] = A[1]/13
print(A, '\n scale Row 2 by 1/13\n')

# replace Row 1 with Row 1 - 2 Row 2
A[0] = A[0] - 2*A[1]
print(A, '\n replace Row 1 with Row 1 - 2 Row 2\n RREF')
````
