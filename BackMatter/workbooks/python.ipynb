4.  Write a function `describe_number(x)` that determines if the number `x` is `"positive"`, `"negative"`, or `"zero"`. Use an `if` statement in your function. Verify your function on the input `4`, `0`, and `-10/3`.

````python
### EXERCISE 4#######################################
#Write your function here

print(describe_number(4), describe_number(0), describe_number(-10/3))
````

<details><summary>Click here for the solution</summary>

```Python
def describe_number(x):
  if x > 0:
    return "positive"
  elif x < 0:
    return "negative"
  else:
    return "zero"

print(describe_number(4), describe_number(0), describe_number(-10/3))
```

5.  Write a function `sum_up_to(n)` that calculates and returns the sum of all whole numbers from `1` up to `n`. Use an `for` statement in your function. Verify your function on the input `1`, `4`, and `100`.

````python
### EXERCISE 5#######################################
#Write your function here

print(sum_up_to(1),sum_up_to(4),sum_up_to(100))
````

<details><summary>Click here for the solution</summary>

```Python
def sum_up_to(n):
  sum = 0
  for i in range(n+1):
    sum += i
  return sum

print(sum_up_to(1),sum_up_to(4),sum_up_to(100))
```

`1 10 5050`
